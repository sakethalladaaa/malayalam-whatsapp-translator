```python
import pandas as pd
import torch

def patched_IndicBERT_roman_inference(
    self,
    IndicLID_BERT_inputs,
    output_dict,
    batch_size
):
    if not IndicLID_BERT_inputs:
        return output_dict

    df = pd.DataFrame(IndicLID_BERT_inputs)
    dataloader = self.get_dataloaders(
        df.iloc[:, 0],
        df.iloc[:, 1],
        batch_size
    )

    with torch.no_grad():
        for data in dataloader:
            batch_indices = data[0]
            batch_inputs = data[1]

            word_embeddings = self.IndicLID_BERT_tokenizer(
                batch_inputs,
                return_tensors="pt",
                padding=True,
                truncation=True,
                max_length=512
            ).to(self.device)

            token_type_ids = word_embeddings.get(
                "token_type_ids",
                torch.zeros_like(word_embeddings["input_ids"])
            )

            batch_outputs = self.IndicLID_BERT(
                word_embeddings["input_ids"],
                token_type_ids=token_type_ids,
                attention_mask=word_embeddings["attention_mask"]
            )

            _, batch_predicted = torch.max(batch_outputs.logits, 1)

            for index, text_input, pred_label, logit in zip(
                batch_indices,
                batch_inputs,
                batch_predicted,
                batch_outputs.logits
            ):
                output_dict[index] = (
                    text_input,
                    self.IndicLID_lang_code_dict_reverse[pred_label.item()],
                    logit[pred_label.item()].item(),
                    "IndicLID-BERT"
                )

    return output_dict


IndicLID.IndicBERT_roman_inference = patched_IndicBERT_roman_inference

print("Tokenizer and BERT inference patch applied successfully.")
```

```python
%%writefile /content/malayalam-whatsapp-translator/phase6/language_detection.py

from pathlib import Path
import os
import sys
from typing import Any


REPO_ROOT = Path(__file__).resolve().parents[1]
INDICLID_ROOT = REPO_ROOT / "IndicLID" / "Inference" / "ai4bharat"

if str(INDICLID_ROOT) not in sys.path:
    sys.path.insert(0, str(INDICLID_ROOT))


class IndicLanguageDetector:
    def __init__(self) -> None:
        self.model = None
        self._load_model()

    def _load_model(self) -> None:
        from IndicLID import IndicLID

        previous_directory = Path.cwd()
        os.chdir(INDICLID_ROOT)

        try:
            self.model = IndicLID(
                input_threshold=0.5,
                roman_lid_threshold=0.6,
            )
        finally:
            os.chdir(previous_directory)

    def predict(self, text: str) -> dict[str, Any]:
        if not text or not text.strip():
            return {
                "text": text,
                "language": None,
                "score": 0.0,
                "model": None,
            }

        result = self.model.batch_predict(
            [text],
            batch_size=1,
        )[0]

        return {
            "text": result[0],
            "language": result[1],
            "score": float(result[2]),
            "model": result[3],
        }


_detector = None


def create_detector() -> IndicLanguageDetector:
    global _detector

    if _detector is None:
        _detector = IndicLanguageDetector()

    return _detector


def detect_language(text: str) -> dict[str, Any]:
    detector = create_detector()
    return detector.predict(text)
```

```python
from pathlib import Path

path = Path("/content/malayalam-whatsapp-translator/phase6/pipeline.py")

code = '''from typing import Any

from phase6.language_detection import detect_language


def process_message(text: str, transliterator=None) -> dict[str, Any]:
    """
    Detect the language of a WhatsApp message and route it through
    the appropriate preprocessing step.

    mal_Latn -> transliterate to Malayalam script
    mal_Mlym -> keep Malayalam script unchanged
    eng_Latn -> keep English unchanged for downstream translation
    unknown  -> keep original text
    """

    detection = detect_language(text)

    if not text or not text.strip():
        return {
            "text": text,
            "language": detection["language"],
            "score": detection["score"],
            "model": detection["model"],
            "processed_text": "",
            "action": "empty",
        }

    language = detection["language"]
    normalized_text = text.strip()

    if language == "mal_Latn":
        if transliterator is None:
            raise RuntimeError(
                "Roman Malayalam detected, but no transliterator was provided."
            )

        processed_text = transliterator(normalized_text)
        action = "transliterate"

    elif language == "mal_Mlym":
        processed_text = normalized_text
        action = "unchanged_malayalam"

    elif language == "eng_Latn":
        processed_text = normalized_text
        action = "english"

    else:
        processed_text = normalized_text
        action = "unsupported"

    return {
        "text": text,
        "language": language,
        "score": detection["score"],
        "model": detection["model"],
        "processed_text": processed_text,
        "action": action,
    }
'''

path.write_text(code, encoding="utf-8")
print(f"Created: {path}")
print(path.read_text(encoding="utf-8"))
```
